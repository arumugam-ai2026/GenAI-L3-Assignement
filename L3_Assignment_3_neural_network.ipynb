```python
import matplotlib.pyplot as plt

# Create figure
fig, ax = plt.subplots(figsize=(12, 8))

# Input Layer (8 Neurons)
for i in range(8):
    ax.scatter(0, i, s=1000, color='skyblue')
    ax.text(-0.4, i, f'X{i+1}', fontsize=10)

# Hidden Layer (6 Neurons)
for i in range(6):
    ax.scatter(4, i + 1, s=1000, color='lightgreen')
    ax.text(4, i + 1, f'H{i+1}',
            ha='center', va='center')

# Output Layer (1 Neuron)
ax.scatter(8, 3, s=1200, color='orange')
ax.text(8, 3, 'Output',
        ha='center', va='center')

# Connections: Input → Hidden
for i in range(8):
    for j in range(6):
        ax.plot([0, 4], [i, j + 1],
                color='gray', linewidth=0.5)

# Connections: Hidden → Output
for i in range(6):
    ax.plot([4, 8], [i + 1, 3],
            color='gray', linewidth=0.5)

# Labels
ax.text(0, 8, "Input Layer", ha='center', fontsize=12)
ax.text(4, 8, "Hidden Layer", ha='center', fontsize=12)
ax.text(8, 8, "Output Layer", ha='center', fontsize=12)

plt.title("Neural Network for Pima Indians Diabetes Prediction")
plt.axis('off')
plt.show()
```
